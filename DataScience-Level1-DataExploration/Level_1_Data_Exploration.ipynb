# Task 1
## Data Exploration and Preprocessing

### Objectives

- Explore the dataset
- Identify rows and columns
- Check missing values
- Handle missing values
- Verify data types
- Analyze Aggregate Rating distribution

```python
# Dataset Shape

print("Number of Rows :", df.shape[0])
print("Number of Columns :", df.shape[1])
```

```python
# Dataset Information

df.info()
```

```python
# Statistical Summary

df.describe().T
```

```python
# Check Missing Values

missing = df.isnull().sum()

missing = missing[missing > 0]

missing.sort_values(ascending=False)
```

```python
# Percentage of Missing Values

missing_percent = (df.isnull().sum()/len(df))*100

missing_percent = missing_percent[missing_percent>0]

missing_percent.sort_values(ascending=False)
```

```python
# Handle Missing Values
# Fill missing cuisines with Unknown

df["Cuisines"] = df["Cuisines"].fillna("Unknown")
```

```python
# Verify Missing Values

df.isnull().sum()
```

```python
# Check Duplicate Records

print("Duplicate Rows :", df.duplicated().sum())
```

```python
# Data Types

df.dtypes
```

```python
# Target Variable Distribution

plt.figure(figsize=(10,6))

sns.histplot(df["Aggregate rating"],
             bins=30,
             kde=True)

plt.title("Distribution of Aggregate Rating")

plt.xlabel("Aggregate Rating")

plt.ylabel("Count")

plt.show()
```

```python
# Rating Categories

plt.figure(figsize=(8,5))

sns.countplot(x="Rating text",
              data=df,
              order=df["Rating text"].value_counts().index)

plt.title("Distribution of Rating Categories")

plt.xticks(rotation=45)

plt.show()
```

```python
# Check Class Imbalance

rating_counts = df["Rating text"].value_counts()

rating_counts
```

### Observations

- The dataset contains **9551 restaurants** and **21 features**.
- Only the **Cuisines** column contains missing values.
- Missing cuisine values were replaced with **"Unknown"**.
- No significant data type issues were found.
- Aggregate ratings are unevenly distributed.
- Many restaurants have ratings between **3.0 and 4.0**.
- A noticeable number of restaurants are marked **Not rated**, indicating class imbalance.

# Task 2
## Descriptive Analysis

This section explores statistical measures, restaurant distribution across cities, countries, and cuisines.

```python
# Numerical Statistics

df.describe().T
```

```python
# Country Distribution

plt.figure(figsize=(10,5))

sns.countplot(y="Country Code",
              data=df,
              order=df["Country Code"].value_counts().index)

plt.title("Restaurants by Country")

plt.show()
```

```python
# City Distribution

top_city = df["City"].value_counts().head(10)

plt.figure(figsize=(12,6))

sns.barplot(x=top_city.values,
            y=top_city.index)

plt.title("Top 10 Cities with Highest Restaurants")

plt.xlabel("Number of Restaurants")

plt.show()
```

```python
# Top 10 Cuisines

cuisines = df["Cuisines"].str.split(", ").explode()

top_cuisine = cuisines.value_counts().head(10)

plt.figure(figsize=(12,6))

sns.barplot(x=top_cuisine.values,
            y=top_cuisine.index)

plt.title("Top 10 Cuisines")

plt.show()
```

```python
# Average Cost Distribution

plt.figure(figsize=(10,6))

sns.histplot(df["Average Cost for two"],
             bins=30)

plt.title("Average Cost for Two")

plt.show()
```

### Observations

- Most restaurants belong to a single country.
- New Delhi has the largest number of restaurants.
- North Indian cuisine is the most common cuisine.
- Restaurant costs are positively skewed, with many restaurants in the lower-to-mid price range.

# Task 3
## Geospatial Analysis

Analyzing restaurant locations using latitude and longitude.

```python
# Interactive Map

fig = px.scatter_mapbox(
    df,
    lat="Latitude",
    lon="Longitude",
    hover_name="Restaurant Name",
    hover_data=["City","Aggregate rating"],
    color="Aggregate rating",
    zoom=2,
    height=700
)

fig.update_layout(mapbox_style="open-street-map")

fig.update_layout(title="Restaurant Locations Across the World")

fig.show()
```

```python
# Restaurants by City

city_count = df.groupby("City").size().sort_values(ascending=False).head(15)

plt.figure(figsize=(12,6))

sns.barplot(x=city_count.values,
            y=city_count.index)

plt.title("Top Cities by Restaurant Count")

plt.show()
```

```python
# Latitude vs Rating

plt.figure(figsize=(10,6))

sns.scatterplot(
    data=df,
    x="Longitude",
    y="Latitude",
    hue="Aggregate rating",
    palette="viridis",
    alpha=0.7
)

plt.title("Restaurant Location vs Aggregate Rating")

plt.show()
```

```python
# Correlation Between Coordinates and Rating

df[["Latitude","Longitude","Aggregate rating"]].corr()
```

### Observations

- The interactive map shows that restaurants are concentrated in a few major cities, indicating higher business activity and customer demand in urban areas.
- The city-wise analysis reveals that some cities have a significantly larger number of restaurants than others, with metropolitan areas dominating the dataset.
- The scatter plot of latitude and longitude colored by aggregate rating suggests that restaurant ratings are distributed across various geographic locations without any obvious spatial pattern.
- The correlation matrix indicates a very weak relationship between geographical coordinates (latitude and longitude) and aggregate rating, suggesting that a restaurant's location alone does not strongly influence its customer rating.
- Overall, geographical location helps identify restaurant distribution and regional density but is not a strong predictor of restaurant ratings.
